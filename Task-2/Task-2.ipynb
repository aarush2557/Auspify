```python
# ==========================================================
# CELL 1: IMPORT LIBRARIES

# Import required libraries
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

# Ignore unnecessary warnings
warnings.filterwarnings("ignore")

# Configure pandas display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)
pd.set_option("display.width", 150)

# Set professional visualization style
plt.style.use("ggplot")
sns.set_style("whitegrid")
sns.set_context("notebook")

# Set default figure parameters
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["figure.dpi"] = 120

print("=" * 70)
print("Netflix Exploratory Data Analysis")
print("Libraries Imported Successfully")
print("=" * 70)
```

```python
# ==========================================================
# CELL 2: LOAD CLEAN DATASET

# Dataset path
DATASET_PATH = Path(
    r"C:\Users\bts\OneDrive\Desktop\Auspify\Dataset\clean_netflix_dataset.csv"
)

try:
    # Load dataset
    df = pd.read_csv(DATASET_PATH)

    print("=" * 70)
    print("Dataset Loaded Successfully")
    print("=" * 70)

    print(f"Dataset Path : {DATASET_PATH}")
    print(f"Rows         : {df.shape[0]}")
    print(f"Columns      : {df.shape[1]}")

    print("\nFirst Five Rows")
    display(df.head())

    print("\nLast Five Rows")
    display(df.tail())

except FileNotFoundError:
    print("Dataset file not found.")
except Exception as error:
    print(f"Error : {error}")
```

```python
# ==========================================================
# CELL 3: DATASET OVERVIEW

print("=" * 70)
print("DATASET OVERVIEW")
print("=" * 70)

# Basic Information
print("\nDataset Information")
df.info()

# Statistical Summary
print("\nNumerical Summary")
display(df.describe())

# Complete Summary
print("\nComplete Summary")
display(df.describe(include="all"))

# Memory Usage
memory_usage = df.memory_usage(deep=True).sum() / (1024 ** 2)

print("\nMemory Usage")
print(f"{memory_usage:.2f} MB")

# Column Names
print("\nColumn Names")

for index, column in enumerate(df.columns, start=1):
    print(f"{index}. {column}")
```

```python
# ==========================================================
# CELL 4: DATA QUALITY CHECK

print("=" * 70)
print("DATA QUALITY CHECK")
print("=" * 70)

# Missing Values
print("\nMissing Values")
missing = df.isnull().sum()

missing_df = pd.DataFrame(
    {
        "Missing Values": missing,
        "Percentage": (missing / len(df) * 100).round(2),
    }
)

display(missing_df)

# Duplicate Records
duplicates = df.duplicated().sum()

print(f"\nDuplicate Rows : {duplicates}")

# Data Types
print("\nData Types")
display(df.dtypes.to_frame(name="Data Type"))

# Unique Values
print("\nUnique Values Per Column")

unique_df = pd.DataFrame(
    {
        "Column": df.columns,
        "Unique Values": [df[col].nunique() for col in df.columns],
    }
)

display(unique_df)

print("=" * 70)
print("Data Quality Check Completed")
print("=" * 70)
```

```python
# ==========================================================
# CELL 5: STATISTICAL SUMMARY

print("=" * 70)
print("STATISTICAL SUMMARY")
print("=" * 70)

# Automatically detect numeric columns
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

if numeric_columns:

    summary = pd.DataFrame(index=numeric_columns)

    summary["Mean"] = df[numeric_columns].mean()
    summary["Median"] = df[numeric_columns].median()

    # Calculate Mode
    summary["Mode"] = [
        df[column].mode().iloc[0]
        if not df[column].mode().empty
        else np.nan
        for column in numeric_columns
    ]

    summary["Standard Deviation"] = df[numeric_columns].std()

    summary["Minimum"] = df[numeric_columns].min()
    summary["25%"] = df[numeric_columns].quantile(0.25)
    summary["50%"] = df[numeric_columns].quantile(0.50)
    summary["75%"] = df[numeric_columns].quantile(0.75)
    summary["Maximum"] = df[numeric_columns].max()

    display(summary.round(2))

else:
    print("No numerical columns found in the dataset.")

print("=" * 70)
print("Statistical Summary Generated Successfully")
print("=" * 70)
```

```python
# ==========================================================
# CELL 6: CONTENT TYPE ANALYSIS

# Analyze the distribution of Movies and TV Shows

print("=" * 70)
print("CONTENT TYPE ANALYSIS")
print("=" * 70)

try:
    # Automatically detect the type column
    possible_columns = ["type", "Type"]

    type_column = next(
        (col for col in possible_columns if col in df.columns),
        None
    )

    if type_column is None:
        raise ValueError("Content type column not found.")

    # Count content types
    type_counts = (
        df[type_column]
        .value_counts()
        .sort_values(ascending=False)
    )

    percentages = (
        type_counts / type_counts.sum() * 100
    ).round(2)

    summary = pd.DataFrame({
        "Count": type_counts,
        "Percentage (%)": percentages
    })

    print("\nContent Distribution")
    display(summary)

    # Count Plot
    plt.figure(figsize=(8, 5))

    ax = sns.countplot(
        data=df,
        x=type_column,
        order=type_counts.index
    )

    plt.title("Netflix Content Distribution")
    plt.xlabel("Content Type")
    plt.ylabel("Count")

    for container in ax.containers:
        ax.bar_label(container)

    plt.tight_layout()
    plt.show()

    # Pie Chart
    plt.figure(figsize=(7, 7))

    plt.pie(
        type_counts,
        labels=type_counts.index,
        autopct="%1.1f%%",
        startangle=90,
        wedgeprops={"edgecolor": "white"}
    )

    plt.title("Movies vs TV Shows")
    plt.tight_layout()
    plt.show()

except Exception as error:
    print(error)
```

```python
# ==========================================================
# CELL 7: RATINGS ANALYSIS

print("=" * 70)
print("RATINGS ANALYSIS")
print("=" * 70)

try:

    rating_column = next(
        (
            col for col in df.columns
            if col.lower() == "rating"
        ),
        None
    )

    if rating_column is None:
        raise ValueError("Rating column not found.")

    rating_counts = (
        df[rating_column]
        .fillna("Unknown")
        .value_counts()
        .head(15)
    )

    print("\nTop Ratings")
    display(rating_counts.to_frame(name="Count"))

    plt.figure(figsize=(12, 6))

    ax = sns.barplot(
        x=rating_counts.index,
        y=rating_counts.values
    )

    plt.title("Top Netflix Ratings")
    plt.xlabel("Rating")
    plt.ylabel("Number of Titles")
    plt.xticks(rotation=45)

    for container in ax.containers:
        ax.bar_label(container)

    plt.tight_layout()
    plt.show()

except Exception as error:
    print(error)
```

```python
# ==========================================================
# CELL 8: TOP COUNTRIES

print("=" * 70)
print("TOP COUNTRIES")
print("=" * 70)

try:

    country_column = next(
        (
            col for col in df.columns
            if col.lower() == "country"
        ),
        None
    )

    if country_column is None:
        raise ValueError("Country column not found.")

    countries = (
        df[country_column]
        .dropna()
        .str.split(",")
        .explode()
        .str.strip()
    )

    top_countries = countries.value_counts().head(10)

    print("\nTop 10 Countries")
    display(top_countries.to_frame(name="Titles"))

    plt.figure(figsize=(10, 6))

    ax = sns.barplot(
        x=top_countries.values,
        y=top_countries.index
    )

    plt.title("Top 10 Countries on Netflix")
    plt.xlabel("Number of Titles")
    plt.ylabel("Country")

    for container in ax.containers:
        ax.bar_label(container)

    plt.tight_layout()
    plt.show()

except Exception as error:
    print(error)
```

```python
# ==========================================================
# CELL 9: TOP DIRECTORS

print("=" * 70)
print("TOP DIRECTORS")
print("=" * 70)

try:

    director_column = next(
        (
            col for col in df.columns
            if col.lower() == "director"
        ),
        None
    )

    if director_column is None:
        raise ValueError("Director column not found.")

    directors = (
        df[director_column]
        .dropna()
        .str.split(",")
        .explode()
        .str.strip()
    )

    top_directors = directors.value_counts().head(10)

    print("\nTop 10 Directors")
    display(top_directors.to_frame(name="Titles"))

    plt.figure(figsize=(12, 6))

    ax = sns.barplot(
        x=top_directors.index,
        y=top_directors.values
    )

    plt.title("Top 10 Directors")
    plt.xlabel("Director")
    plt.ylabel("Titles")
    plt.xticks(rotation=45, ha="right")

    for container in ax.containers:
        ax.bar_label(container)

    plt.tight_layout()
    plt.show()

except Exception as error:
    print(error)
```

```python
# ==========================================================
# CELL 10: TOP ACTORS

print("=" * 70)
print("TOP ACTORS")
print("=" * 70)

try:

    cast_column = next(
        (
            col for col in df.columns
            if col.lower() in ["cast", "actors"]
        ),
        None
    )

    if cast_column is None:
        raise ValueError("Cast column not found.")

    actors = (
        df[cast_column]
        .dropna()
        .str.split(",")
        .explode()
        .str.strip()
    )

    top_actors = actors.value_counts().head(15)

    print("\nTop 15 Actors")
    display(top_actors.to_frame(name="Appearances"))

    plt.figure(figsize=(11, 7))

    ax = sns.barplot(
        x=top_actors.values,
        y=top_actors.index
    )

    plt.title("Top 15 Most Featured Actors")
    plt.xlabel("Appearances")
    plt.ylabel("Actor")

    for container in ax.containers:
        ax.bar_label(container)

    plt.tight_layout()
    plt.show()

except Exception as error:
    print(error)
```

```python
# ==========================================================
# CELL 11: GENRE ANALYSIS

# Analyze the 'listed_in' column to identify the most popular genres

print("=" * 70)
print("GENRE ANALYSIS")
print("=" * 70)

try:
    # Automatically detect genre column
    genre_column = next(
        (col for col in df.columns if col.lower() == "listed_in"),
        None
    )

    if genre_column is None:
        raise ValueError("Genre column ('listed_in') not found.")

    # Split multiple genres and count occurrences
    genres = (
        df[genre_column]
        .dropna()
        .str.split(",")
        .explode()
        .str.strip()
    )

    top_genres = genres.value_counts().head(15)

    print("\nTop 15 Genres")
    display(top_genres.to_frame(name="Titles"))

    # Professional Bar Plot
    plt.figure(figsize=(12, 7))

    ax = sns.barplot(
        x=top_genres.values,
        y=top_genres.index
    )

    plt.title("Top 15 Netflix Genres", fontsize=16, weight="bold")
    plt.xlabel("Number of Titles")
    plt.ylabel("Genre")

    for container in ax.containers:
        ax.bar_label(container)

    plt.tight_layout()
    plt.show()

except Exception as error:
    print(f"Error: {error}")
```

```python
# ==========================================================
# CELL 12: RELEASE YEAR ANALYSIS

# Analyze release trends across years

print("=" * 70)
print("RELEASE YEAR ANALYSIS")
print("=" * 70)

try:
    release_column = next(
        (col for col in df.columns if col.lower() == "release_year"),
        None
    )

    if release_column is None:
        raise ValueError("Release year column not found.")

    release_years = (
        df[release_column]
        .dropna()
        .astype(int)
    )

    yearly_counts = (
        release_years
        .value_counts()
        .sort_index()
    )

    print("\nRelease Year Summary")
    display(yearly_counts.tail(15))

    # Line Plot
    plt.figure(figsize=(14, 6))

    plt.plot(
        yearly_counts.index,
        yearly_counts.values,
        linewidth=2
    )

    plt.title("Netflix Content Released Over the Years",
              fontsize=16,
              weight="bold")
    plt.xlabel("Release Year")
    plt.ylabel("Number of Titles")

    plt.tight_layout()
    plt.show()

    # Histogram
    plt.figure(figsize=(10, 6))

    sns.histplot(
        release_years,
        bins=30,
        kde=True
    )

    plt.title("Distribution of Release Years",
              fontsize=16,
              weight="bold")

    plt.xlabel("Release Year")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()

except Exception as error:
    print(error)
```

```python
# ==========================================================
# CELL 13: CONTENT ADDED OVER TIME

# Analyze trends based on 'date_added'

print("=" * 70)
print("CONTENT ADDED OVER TIME")
print("=" * 70)

try:
    date_column = next(
        (col for col in df.columns if col.lower() == "date_added"),
        None
    )

    if date_column is None:
        raise ValueError("date_added column not found.")

    # Convert to datetime
    dates = pd.to_datetime(
        df[date_column],
        errors="coerce"
    )

    # Monthly Trend
    monthly = (
        dates.dt.month_name()
        .value_counts()
        .reindex([
            "January", "February", "March",
            "April", "May", "June",
            "July", "August", "September",
            "October", "November", "December"
        ])
    )

    # Yearly Trend
    yearly = (
        dates.dt.year
        .value_counts()
        .sort_index()
    )

    # Monthly Plot
    plt.figure(figsize=(13, 5))

    sns.lineplot(
        x=monthly.index,
        y=monthly.values,
        marker="o"
    )

    plt.xticks(rotation=45)
    plt.title("Monthly Content Addition Trend",
              fontsize=15,
              weight="bold")

    plt.xlabel("Month")
    plt.ylabel("Titles Added")

    plt.tight_layout()
    plt.show()

    # Yearly Plot
    plt.figure(figsize=(13, 5))

    sns.lineplot(
        x=yearly.index,
        y=yearly.values,
        marker="o"
    )

    plt.title("Yearly Content Addition Trend",
              fontsize=15,
              weight="bold")

    plt.xlabel("Year")
    plt.ylabel("Titles Added")

    plt.tight_layout()
    plt.show()

except Exception as error:
    print(error)
```

```python
# ==========================================================
# CELL 14: MOVIE DURATION ANALYSIS

# Analyze duration separately for Movies and TV Shows

print("=" * 70)
print("MOVIE DURATION ANALYSIS")
print("=" * 70)

try:

    duration_column = next(
        (col for col in df.columns if col.lower() == "duration"),
        None
    )

    type_column = next(
        (col for col in df.columns if col.lower() == "type"),
        None
    )

    if duration_column is None or type_column is None:
        raise ValueError("Required columns not found.")

    duration_df = df[[type_column, duration_column]].copy()

    # Extract numeric duration
    duration_df["Duration_Value"] = (
        duration_df[duration_column]
        .str.extract(r"(\d+)")
        .astype(float)
    )

    display(duration_df.head())

    # Histogram
    plt.figure(figsize=(12, 6))

    sns.histplot(
        data=duration_df,
        x="Duration_Value",
        hue=type_column,
        bins=30,
        kde=True
    )

    plt.title("Distribution of Content Duration",
              fontsize=16,
              weight="bold")

    plt.xlabel("Duration")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()

    # Boxplot
    plt.figure(figsize=(8, 6))

    sns.boxplot(
        data=duration_df,
        x=type_column,
        y="Duration_Value"
    )

    plt.title("Duration Comparison",
              fontsize=16,
              weight="bold")

    plt.tight_layout()
    plt.show()

except Exception as error:
    print(error)
```

```python
# ==========================================================
# CELL 15: CORRELATION ANALYSIS

# Generate correlation matrix for numerical columns

print("=" * 70)
print("CORRELATION ANALYSIS")
print("=" * 70)

try:

    numeric_df = df.select_dtypes(include=np.number)

    if numeric_df.empty:
        print("No numerical columns available for correlation.")

    else:

        correlation = numeric_df.corr()

        print("\nCorrelation Matrix")
        display(correlation)

        plt.figure(figsize=(10, 8))

        sns.heatmap(
            correlation,
            annot=True,
            cmap="coolwarm",
            linewidths=0.5,
            fmt=".2f"
        )

        plt.title(
            "Correlation Heatmap",
            fontsize=16,
            weight="bold"
        )

        plt.tight_layout()
        plt.show()

except Exception as error:
    print(error)
```

```python
# ==========================================================
# CELL 16: BUSINESS INSIGHTS

# Generate business insights automatically from the dataset

print("=" * 80)
print("BUSINESS INSIGHTS")
print("=" * 80)

try:

    # Top Country
    top_country = (
        df["country"]
        .dropna()
        .str.split(",")
        .explode()
        .str.strip()
        .value_counts()
        .idxmax()
    )

    # Most Common Genre
    top_genre = (
        df["listed_in"]
        .dropna()
        .str.split(",")
        .explode()
        .str.strip()
        .value_counts()
        .idxmax()
    )

    # Most Frequent Rating
    top_rating = df["rating"].mode()[0]

    # Movie vs TV Ratio
    type_counts = df["type"].value_counts()

    movie_count = type_counts.get("Movie", 0)
    tv_count = type_counts.get("TV Show", 0)

    ratio = (
        movie_count / tv_count
        if tv_count != 0
        else np.nan
    )

    # Release Years
    newest = int(df["release_year"].max())
    oldest = int(df["release_year"].min())

    print(f"Top Country            : {top_country}")
    print(f"Most Common Genre      : {top_genre}")
    print(f"Most Frequent Rating   : {top_rating}")
    print(f"Movies                 : {movie_count}")
    print(f"TV Shows               : {tv_count}")
    print(f"Movie : TV Ratio       : {ratio:.2f}")
    print(f"Newest Content         : {newest}")
    print(f"Oldest Content         : {oldest}")

except Exception as error:
    print(error)
```

```python
# ==========================================================
# CELL 17: ADVANCED VISUALIZATIONS

print("=" * 80)
print("ADVANCED VISUALIZATIONS")
print("=" * 80)

# Professional Visualization 1
plt.figure(figsize=(10, 6))
sns.countplot(data=df, y="rating",
              order=df["rating"].value_counts().index)
plt.title("Rating Distribution")
plt.tight_layout()
plt.show()

# Professional Visualization 2
plt.figure(figsize=(12, 6))
sns.boxplot(data=df,
            x="type",
            y=df["duration"].str.extract(r"(\d+)").astype(float)[0])
plt.title("Duration Comparison")
plt.tight_layout()
plt.show()

# Professional Visualization 3
plt.figure(figsize=(10, 6))
sns.histplot(df["release_year"],
             bins=25,
             kde=True)
plt.title("Release Year Distribution")
plt.tight_layout()
plt.show()

# Professional Visualization 4
plt.figure(figsize=(12, 7))
sns.countplot(
    y=df["listed_in"]
    .str.split(",")
    .explode()
    .str.strip(),
    order=df["listed_in"]
    .str.split(",")
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
    .index
)
plt.title("Top 10 Genres")
plt.tight_layout()
plt.show()
```
